```python
# Problema 1. Clasificador de temperatura

# Recorremos los números del -30 al 50
for temperatura in range(-30,51):
# Si el número es menor que 0, imprime "Muy frío"
 if temperatura < 0:
  print(temperatura,"Muy frío")

# Si el número está entre 0 y 24, imprime "Frío"
 elif temperatura >= 0 and temperatura <= 24:
  print(temperatura,"Frío")

# Si el número es igual o mayor a 25, imprime "Calor"
 else:
  print(temperatura,"Calor")
```

```python
# Problema 2. Números especiales del robot

# Recorremos los números del 1 al 100
for número in range(1,101):
# Si número es divisible por 10, imprime "Bip Bip"
  if número % 10 == 0:
    print(número,"Bip Bip")

# Si no lo es y es divisible por 5, imprime "Bip"
  elif número % 5 == 0:
    print(número,"Bip")
```

```python
# Problema 3. Detector de puntaje gamer

# Solicitamos dos números al usuario
puntaje1 = int(input("Ingrese el primer puntaje: "))
puntaje2 = int(input("Ingresa el segundo puntaje: "))

# Sumamos ambos números
suma = puntaje1 + puntaje2

# Mostramos el mensaje según los resultados
if suma < 50:
  print("Nivel principiante:",suma)
elif suma < 80:
  print("Nivel intermedio:",suma)
else:
  print("Nivel pro:",suma)


```
